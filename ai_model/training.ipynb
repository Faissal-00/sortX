```python
from ultralytics import YOLO

# 1. Load the pre-trained Nano model (the fastest one, perfect for laptops)
model = YOLO("yolov8n.pt")

# 2. Start the training process
print("Starting SortX AI training on the Quadro GPU... Let's go!")
results = model.train(
    data=r"C:\Users\hp zbook\Desktop\straiii\ai_model\tomato ripeness.v2i.yolov8\data.yaml",
    epochs=100,        # 15 rounds is enough for a quick test today
    imgsz=640,        # Standard image size for YOLO
    batch=8,          # Processing 8 images at a time (keeps your laptop from freezing)
    name="sortx_v1",  # This creates a specific folder for your results
    device=0,         # ✅ Forces YOLO to use your Quadro T1000 GPU (GPU 0)
    patience=10        # ✅ Stops training early if it doesn't improve for 5 rounds
)
```
